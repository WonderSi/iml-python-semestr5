№2 (4 балла) <br/>
Создайте рисунок с основными элементами изображения, похожий на приведенный ниже.
Элементы изображения обозначены кругами и подписаны синим шрифтом. Допускается
изменение линий, цвета, толщины линий и т.п., главное, чтобы все обозначенные элементы
присутствовали.

```python
np.random.seed(42)
x_scatter = np.random.uniform(0.5, 3.5, 50)
y_scatter = np.random.uniform(0.2, 3.5, 50)
plt.scatter(x_scatter, y_scatter, color='white', s=30, alpha=0.6, marker='o', 
           facecolors='none', edgecolors='black')

x_line = np.linspace(0.5, 3.8, 100)
y_line = 3 * np.exp(-2 * (x_line - 1)) + 2
plt.plot(x_line, y_line, 'b-', linewidth=2, label='Синяя линия')

x_spline = np.linspace(0.5, 3.5, 100)
y_spline = 0.5 + 0.3 * np.sin(2 * (x_spline - 1)) + 0.2 * (x_spline - 1)
plt.plot(x_spline, y_spline, 'r-', linewidth=2, label='Красная линия')

plt.xlabel('Подпись оси OX')
plt.ylabel('Подпись оси OY')

plt.gca().yaxis.set_major_locator(plt.MultipleLocator(1.0))
plt.gca().yaxis.set_minor_locator(plt.MultipleLocator(0.25))

plt.gca().xaxis.set_major_locator(plt.MultipleLocator(1.0))
plt.gca().xaxis.set_minor_locator(plt.MultipleLocator(0.25))
plt.gca().xaxis.set_minor_formatter(plt.FormatStrFormatter('%.2f'))

plt.grid(True, which='major', linestyle='-', alpha=0.7)
plt.grid(True, which='minor', linestyle='--', alpha=0.4)

plt.title('Элементы изображения')
plt.legend()

plt.xlim(0, 4)
plt.ylim(0, 4)

plt.show()
```

№3 (2 балла) <br/>
Постройте график функции <br/>
 <br/>
и по графику найдите корни уравнения y(x) = 0. <br/>
(Не нужно применять численных методов — просто так подберите отображение, чтобы на графике были видны корни функции.)

```python
x = np.linspace(-3, 4, 1000)
y = x**2 - x - 6

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(x, y, label="y(x) = x^2 - x - 6", color="blue", linewidth=2)

ax.axhline(0, color='gray', linestyle='--', linewidth=1)

roots = [-2, 3]
ax.scatter(roots, [0, 0], color='red', zorder=5, label='Roots $y=0$')

ax.set_title("y(x) = x^2 - x - 6")
ax.set_xlabel("x")
ax.set_ylabel("y")
ax.legend()
ax.grid(True)

plt.show()
```

№4 (2 балла) <br/>
Постройте график функции <br/>


```python

def calc(t):
    a = 1 + np.tan(1 / (1 + np.sin(t)**2))
    b = (t**2 + 1) * np.exp(-np.abs(t) / 10)
    res = np.log(b) / np.log(a)
    res[(a <= 0) | (a == 1)] = np.nan
    return res

t = np.linspace(-10, 10, 1000)
plt.plot(t, calc(t), 'blue', linewidth=2)
plt.grid(True)
plt.ylim(-2, 11)
plt.xlim(-11, 11)
plt.show()
```
